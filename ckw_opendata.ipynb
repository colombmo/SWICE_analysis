```python
import pandas as pd
```

```python
df = pd.read_csv('data/ckw_opendata_smartmeter_dataset_a_202601.csv')
```

```python
df.head()
```

```python
# Count number of unique ids
num_unique_ids = df['id'].nunique()
print(f'Number of households: {num_unique_ids}')
```

```python
# Get number of occurrences for each id
id_counts = df['id'].value_counts()
print(id_counts)
```

```python
# Keep only ids with more than 2500 occurrences
ids_to_keep = id_counts[id_counts > 2500].index
df_filtered = df[df['id'].isin(ids_to_keep)]
```

```python
# Number of households left after filtering
num_filtered_ids = df_filtered['id'].nunique()
print(f'Number of households after filtering: {num_filtered_ids}')
# NUmber of occurrences for each id after filtering
filtered_id_counts = df_filtered['id'].value_counts()
print(filtered_id_counts)
```

```python
# COnvert the timestamp column to datetime format
df_filtered['timestamp'] = pd.to_datetime(df_filtered['timestamp'])
```

```python
# Save as compressed pickle file
df_filtered.to_pickle('data/ckw_opendata_smartmeter_dataset_a_202601_filtered.pkl.gz', compression='gzip')
```
